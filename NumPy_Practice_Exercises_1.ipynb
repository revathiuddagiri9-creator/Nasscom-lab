### Q13. Why NumPy is faster than plain Python
Create two Python lists of 500,000 numbers each. Time how long it takes to add them element-by-element using a Python loop (or list comprehension). Then convert both lists to NumPy arrays and time how long it takes to add them using `+`. Print both times and, in a comment, state which was faster and roughly by how much.

```python
# Write your code here
import numpy as np
import time

a = list(range(500000))
b = list(range(500000))

t = time.time()
[a[i] + b[i] for i in range(500000)]
p = time.time() - t

a, b = np.array(a), np.array(b)

t = time.time()
a + b
n = time.time() - t

print("Python:", p)
print("NumPy:", n)
# NumPy was faster by roughly p/n times.
```

### Q16. Vectorized operations and boolean masking
Create a 1-D NumPy array of 8 exam scores (0-100) of your choice. Without using any loop:
1. Add 5 bonus marks to every score and print the result.
2. Print a boolean array showing which scores are 50 or above.
3. Print only the scores that are 50 or above using boolean masking.

```python
# Write your code here
import numpy as np

scores = np.array([45, 67, 82, 39, 55, 91, 48, 73])

bonus_scores = scores + 5
print(bonus_scores)

passed = bonus_scores >= 50
print(passed)

print(bonus_scores[passed])


```
